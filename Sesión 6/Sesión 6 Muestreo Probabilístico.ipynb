# Muestreo Sistemático

Ejemplo 1: Supongamos que un experto en seguros trabaja para una aseguradora de vida y está interesado en estudiar la duración promedio de las pólizas de seguro de vida que se han vendido durante el último año. La compañía cuenta con un total de 1500 pólizas y debe seleccionar una muestra representativa.

1. Generamos una base de datos para aplicar esta técnica de muestreo.

```python


np.random.seed(123)
NOMBRES_EJEMPLO = ['Olivia', 'Emma', 'Charlotte', 'Amelia', 'Sophia', 'Isabella', 'Ava', 'Mia', 'Evelyn', 'Luna']
base_asegurados = pd.DataFrame({
    'Nombre': np.random.choice(NOMBRES_EJEMPLO, size=1500, replace=True),
    'Duración': np.round(np.random.uniform(low=1, high=12, size=1500))
})
print(base_asegurados.head())
```

2. Seleccionamos una muestra de tamaño 300.

```python
N_poblacional = 1500
n_muestral = 300
ancho_intervalo = int(round(N_poblacional / n_muestral))
np.random.seed(123)
primer_elemento = np.random.choice(range(1, ancho_intervalo + 1), size=1)[0]
secuencia = np.arange(start=primer_elemento, stop=N_poblacional, step=ancho_intervalo)
base_muestra_asegurados = base_asegurados.iloc[secuencia]
print(base_muestra_asegurados.head(10))
```

3. Calculamos la duración promedio de las pólizas

```python
promedio_muestra = base_muestra_asegurados['Duración'].mean()
promedio_poblacional = base_asegurados['Duración'].mean()
print(f'Promedio Muestra: {promedio_muestra}')
print(f'Promedio Poblacional: {promedio_poblacional}')
```

Ejemplo 2: Una universidad está interesada en evaluar el desempeño promedio de sus estudiantes en un examen estandarizado aplicado durante el último semestre. La universidad cuenta con un total de 1500 estudiantes que realizaron el examen. Para ahorrar tiempo y recursos, se ha decidido realizar un análisis sobre una muestra representativa de 300 estudiantes.

1. Generamos una base de datos para practicar.

```python
np.random.seed(234)
base_estudiantes = pd.DataFrame({
    'Nombre': np.random.choice(NOMBRES_EJEMPLO, size=1500, replace=True),
    'Calificación': np.round(np.random.uniform(low=4, high=10, size=1500))
})
print(base_estudiantes.head())
```

2. Seleccionamos una muestra de tamaño 300.

```python
N_estudiantes = 1500
n_estudiantes = 175
ancho_intervalo = int(np.ceil(N_estudiantes / n_estudiantes))
np.random.seed(123)
primer_elemento = np.random.choice(range(1, ancho_intervalo + 1), size=1)[0]
secuencia = np.arange(start=primer_elemento, stop=N_estudiantes, step=ancho_intervalo)
n_faltantes = n_estudiantes - len(secuencia)
if n_faltantes > 0:
    indices_no_seleccionados = np.setdiff1d(range(N_estudiantes), secuencia)
    indices_adicionales = np.random.choice(indices_no_seleccionados, size=n_faltantes, replace=False)
    secuencia = np.concatenate([secuencia, indices_adicionales])
base_muestra_estudiantes = base_estudiantes.iloc[secuencia]
print(base_muestra_estudiantes.head(10))
```

3. Calculamos la calificación promedio.

```python
print(base_muestra_estudiantes['Calificación'].mean())
print(base_estudiantes['Calificación'].mean())
```

# Muestreo Estratificado

Ejemplo 3: Trabajas para una compañía de seguros de automóviles y estás interesado en estimar la proporción de asegurados que han presentado reclamos por accidentes de tráfico en diferentes regiones geográficas de México.

1. Generamos una base de datos para practicar

```python
np.random.seed(123)
cartera_asegurados = pd.DataFrame({
    'Nombre': np.random.choice(NOMBRES_EJEMPLO, size=1000, replace=False),
    'Region': np.random.choice(['Norte', 'Centro', 'Sur'], size=1000, replace=True),
    'Reclamación': np.random.choice(['Sí', 'No'], size=1000, replace=True)
})
print(cartera_asegurados.head())
```

2. Seleccionamos una muestra de tamaño 200 a través de un muestreo estratificado

2.1. La muestra es igual en todos los estratos (por ejemplo: 67, 67 y 66)

```python
print(cartera_asegurados[''Region''].value_counts())
np.random.seed(567)
# Aseguramos que el tamaño de la muestra no exceda el tamaño del grupo
def sample_group(x):
    n_sample = 67 if x.name != 'Sur' else 66
    return x.sample(n=min(n_sample, len(x)))

asegurados_muestra = cartera_asegurados.groupby('Region').apply(sample_group).reset_index(drop=True)
print(asegurados_muestra)
```

Calculamos la proporción de reclamaciones por región

```python
print(pd.crosstab(asegurados_muestra[''Region''], asegurados_muestra[''Reclamación''], normalize='all'))
print(pd.crosstab(cartera_asegurados[''Region''], cartera_asegurados[''Reclamación''], normalize='all'))
```

2.2. La muestra es proporcional al tamaño de los estratos.

```python
N = 1000
n = 200
proporciones = cartera_asegurados[''Region''].value_counts(normalize=True)
tamaños_muestra = round(proporciones * n).astype(int)
print(tamaños_muestra)
np.random.seed(567)

def sample_proportional(x):
    n_sample = tamaños_muestra[x.name]
    return x.sample(n=min(n_sample, len(x)))
asegurados_muestra = cartera_asegurados.groupby('Region').apply(sample_proportional).reset_index(drop=True)
print(asegurados_muestra.head())
```

Calculamos la proporción de reclamaciones por región

```python
print(pd.crosstab(asegurados_muestra[''Region''], asegurados_muestra[''Reclamación''], normalize='all'))
print(pd.crosstab(cartera_asegurados[''Region''], cartera_asegurados[''Reclamación''], normalize='all'))
```

Forma alternaativa: Esta forma sirve cuando los estratos son "muchos"

```python
np.random.seed(123)
tamaños_muestra = cartera_asegurados.groupby('Region').size().reset_index(name='counts').assign(
    Prop=lambda x: x['counts'] / N,
    Tamaño_Muestra=lambda x: round(x['Prop'] * 200).astype(int)
)
asegurados_muestra = cartera_asegurados.groupby('Region').apply(lambda x: x.sample(n=tamaños_muestra.loc[tamaños_muestra.Region==x.name, ''Tamaño_Muestra''].iloc[0])).reset_index(drop=True)
print(asegurados_muestra.head())
print(pd.crosstab(asegurados_muestra[''Region''], asegurados_muestra[''Reclamación''], normalize='all'))
print(pd.crosstab(cartera_asegurados[''Region''], cartera_asegurados[''Reclamación''], normalize='all'))
```

Problema: Una compañía de seguros de automóviles está interesada en analizar los patrones de reclamos realizados por sus clientes, desglosados por el día de la semana en que ocurrieron. La base de datos anual contiene información detallada sobre los clientes, las características de sus pólizas, y los incidentes reportados. Dentro de esta base de datos, la variable DayOfWeek identifica el día de la semana en que ocurrieron los incidentes.

1. Calcular las proporciones por día de la semana
Determina el porcentaje de incidentes que ocurrieron en cada día de la semana respecto al total de registros.

2. Determinar el tamaño de muestra proporcional
Calcula el tamaño de la muestra para cada día de la semana, asegurando que la muestra total sea de 500 registros.

3. Seleccionar una muestra aleatoria proporcional
Utiliza las proporciones calculadas para seleccionar de manera aleatoria los registros correspondientes a cada día de la semana.

```python
fraudes = pd.read_excel("fraud_oracle.xlsx")
n_muestra_total = 500
muestra = fraudes.groupby('DayOfWeek', group_keys=False).apply(lambda x: x.sample(frac=n_muestra_total/len(fraudes), random_state=123))
print(muestra.head())
```
